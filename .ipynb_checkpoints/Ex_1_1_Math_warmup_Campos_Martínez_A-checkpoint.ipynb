```python
# Ejercicio_1
#Clase Math_warmup
#Agustin Jaime Navarro

import numpy as np

def linear_1d_student(x, beta, omega):
    """
    Calcula la función lineal y = beta + omega * x.
    x puede ser un escalar o un array numpy; la operación es vectorizada.
    """
    x = np.asarray(x)
    y = beta + omega * x
    return y

# Pruebas rápidas (no modificar)
x_test = np.array([0.0, 1.0, 2.0])
y_test = linear_1d_student(x_test, beta=2.0, omega=3.0)
assert np.allclose(y_test, np.array([2.0, 5.0, 8.0]))
print("All tests passed ✅")

# Ejemplos adicionales
if __name__ == "__main__":
    # Escalar
    print("y(scalar) =", linear_1d_student(4.0, beta=1.0, omega=0.5))  # 1 + 0.5*4 = 3.0

    # Array
    xs = np.linspace(-1, 1, 5)
    print("x:", xs)
    print("y:", linear_1d_student(xs, beta=0.0, omega=2.0))  # y = 2*x
```
